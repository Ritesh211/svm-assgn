## **Question 1:** What is Information Gain, and how is it used in Decision Trees?

**Answer:**  
Information Gain (IG) measures how much “information” a feature gives about the class.  
It quantifies the reduction in entropy after splitting the data based on a feature.

**Formula:**
IG(S, A) = Entropy(S) - Σ (|Sv| / |S| × Entropy(Sv))

Decision Trees use IG to select the feature that best separates the data — i.e., the one with the **highest Information Gain**.

## **Question 2:** What is the difference between Gini Impurity and Entropy?

| Measure | Formula | Range | When to Use | Key Difference |
|----------|----------|--------|--------------|----------------|
| **Entropy** | `-Σ p_i log2(p_i)` | 0–1 | Used in ID3, C4.5 | Based on Information Theory |
| **Gini Impurity** | `1 - Σ p_i²` | 0–0.5 | Used in CART | Measures probability of misclassification |

- **Entropy** is more computationally intensive due to log operations.  
- **Gini** is faster and often gives similar results.  
- **Both** aim to find the most “pure” node split.

## **Question 3:** What is Pre-Pruning in Decision Trees?

**Answer:**  
Pre-pruning (early stopping) halts the tree’s growth before it perfectly classifies the training data.  
It prevents **overfitting** by setting constraints such as:
- Maximum depth (`max_depth`)
- Minimum samples per split (`min_samples_split`)
- Minimum impurity decrease

It ensures a simpler, more generalizable model.

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

# Load data
data = load_iris()
X, y = data.data, data.target

# Train Decision Tree using Gini
clf = DecisionTreeClassifier(criterion='gini', random_state=42)
clf.fit(X, y)

# Feature importance
for name, importance in zip(data.feature_names, clf.feature_importances_):
    print(f"{name}: {importance:.3f}")
Sample Output:

java
Copy code
petal length (cm): 0.550
petal width (cm): 0.420
sepal length (cm): 0.025
sepal width (cm): 0.005
Question 5: What is a Support Vector Machine (SVM)?
Answer:
SVM is a supervised learning algorithm used for classification and regression.
It finds the optimal hyperplane that maximizes the margin between classes in a high-dimensional space.

Key idea:

ini
Copy code
Margin = 2 / ||w||
where w is the weight vector perpendicular to the hyperplane.

Question 6: What is the Kernel Trick in SVM?
Answer:
The Kernel Trick allows SVMs to work efficiently in non-linear spaces by mapping data into higher dimensions without explicitly computing coordinates.

Common kernels:

Linear: simple dot product

Polynomial: (x·y + c)^d

RBF (Gaussian): exp(-γ ||x - y||²)

This makes SVM powerful for complex datasets.

Question 7: Python Program — SVM with Linear & RBF Kernels (Wine Dataset)
python
Copy code
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load dataset
X, y = load_wine(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Linear kernel
svm_linear = SVC(kernel='linear', random_state=42)
svm_linear.fit(X_train, y_train)

# RBF kernel
svm_rbf = SVC(kernel='rbf', random_state=42)
svm_rbf.fit(X_train, y_train)

# Compare accuracies
print("Linear Kernel Accuracy:", accuracy_score(y_test, svm_linear.predict(X_test)))
print("RBF Kernel Accuracy:", accuracy_score(y_test, svm_rbf.predict(X_test)))
Sample Output:

yaml
Copy code
Linear Kernel Accuracy: 0.98
RBF Kernel Accuracy: 1.00
Question 8: What is the Naïve Bayes classifier, and why is it called "Naïve"?
Answer:
Naïve Bayes is a probabilistic classifier based on Bayes’ Theorem:

mathematica
Copy code
P(C|X) = [P(X|C) * P(C)] / P(X)
It assumes that features are independent given the class — hence “naïve.”
Despite this assumption, it performs well in many real-world tasks like spam detection and text classification.

Question 9: Differences among Naïve Bayes Types
answer--
| Type               | Used For            | Assumption                    | Example Use            |
| ------------------ | ------------------- | ----------------------------- | ---------------------- |
| **Gaussian NB**    | Continuous features | Normally distributed features | Medical or sensor data |
| **Multinomial NB** | Discrete counts     | Features represent counts     | Text classification    |
| **Bernoulli NB**   | Binary features     | Features are 0/1              | Spam detection         |


Question 10: Python Program — Gaussian Naïve Bayes (Breast Cancer Dataset)
python
Copy code
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load dataset
data = load_breast_cancer()
X, y = data.data, data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train model
model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))
Sample Output:

makefile
Copy code
Accuracy: 0.95
